# Zomato Data Cleaning and Feature Engineering
This notebook performs data ingestion, cleaning, and preparation of the Zomato restaurant dataset.

```python
# Import the pandas library, commonly aliased as 'pd', for data manipulation and analysis.
import pandas as pd
```

## Step 0: Data Ingestion and Initial Exploration

```python
# Attempt to read the 'zomato.csv' file into a pandas DataFrame using UTF-8 encoding.
try:
    df = pd.read_csv('zomato.csv', encoding='utf-8')
# If UTF-8 fails (due to a UnicodeDecodeError), try reading the file again using 'latin1' encoding.
except UnicodeDecodeError:
    df = pd.read_csv('zomato.csv', encoding='latin1')
# Generate descriptive statistics of the DataFrame, showing central tendency, dispersion, and shape of its distribution.
df.describe()
```

```python
# Print a concise summary of the DataFrame, including the data types of columns, non-null values, and memory usage.
df.info()
```

```python
# Display the names of all columns in the DataFrame.
df.columns
```

## Preliminary Missing Value Analysis

```python
# Calculate the number of missing (NaN) values for each column in the DataFrame.
df.isna().sum()
```

## Step 1: Fill Missing Values in Cuisines

```python
# Fill any missing values (NaN) in the 'Cuisines' column with the string 'unknown'.
df['Cuisines']=df['Cuisines'].fillna('unknown')
# Print the total count of missing values in the 'Cuisines' column after filling them.
print("Missing values:", df['Cuisines'].isnull().sum())
```

### Data Verification - Post Step 1

```python
df.head(6)
```

## Step 2: Drop Redundant & Zero-Variance Columns

```python
# Step 2: Drop redundant and zero-variance columns
cols_to_drop = ['Switch to order menu', 'Locality Verbose', 'Address']
df = df.drop(columns=cols_to_drop, errors='ignore')

# Verify remaining columns and new shape
print("Current Shape:", df.shape)
print("\nRemaining Columns:\n", df.columns.tolist())
```

### Data Verification - Post Step 2

```python
df.head(6)
```

## Step 3: Created new column and convert Binary Yes/No Text Columns to 1 / 0 Numerical Flags

```python
# Define a list of binary categorical column names that need to be converted.
binary_cols=['Has Table booking','Has Online delivery','Is delivering now']
# Iterate through each column in the binary_cols list.
for cols in binary_cols:
  # Create new columns with '_flags' suffix. Map 'Yes' to '1' and 'No' to '0' for numerical representation.
  df[cols + ' flags']=df[cols].map({"Yes":"1", "No":"0"})
# Print the head of the DataFrame, showing the original and newly created 'flags' columns for verification.
print(df[['Has Table booking','Has Table booking flags','Has Online delivery','Has Online delivery flags','Is delivering now', 'Is delivering now flags']].head(7))
```

### Verify Flag Creation

```python
#checking the numeric flag columns
df.columns
```

## Step 4: Feature Engineering — Flag Unrated Restaurants

```python
# checking rating status newly open restuarant
df[df['Aggregate rating']==0].head(6)
```

```python
# Create a flag for Rated vs. Unrated Restaurants
df['is rated']=(df['Aggregate rating'] > 0).astype(int)
# Verify the newly engineered features
df[['Aggregate rating','is rated']].head(20)
```

### Verification of Engineered Features

```python
# final checking of the columns
df.columns
```

```python
df.info()
```

### Dropping some more columns that is not necessary

```python
# Drop the Primary_Cuisine column
df = df.drop(columns=['Primary_Cuisine'], errors='ignore')
df.columns.to_list()
```

```python
# droping 'rating color' column.
df = df.drop(columns=['Rating color'], errors='ignore')
df.columns
```

##Step 5: Standardizing column names to snake_case (lowercase with underscores)

```python
# Convert all column names to lowercase and replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')
print("the final columns list: ", df.columns.to_list())
```
